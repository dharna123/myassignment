```python
bert_preprocess=hub.KerasLayer("https://kaggle.com/models/tensorflow/bert/TensorFlow2/en-uncased-preprocess/3")
```

```python
bert_encoder=hub.KerasLayer("https://www.kaggle.com/models/tensorflow/bert/TensorFlow2/en-uncased-l-12-h-768-a-12/4")
```

```python
import tensorflow_text as text
```

```python
!pip install tensorflow_text
```

```python
def get_sentence_embeding(sentences):
    preprocessed_text=bert_preprocess([sentences])
    return bert_encoder(preprocessed_text)['pooled_output'][0]
```

```python
get_sentence_embeding(["Free discount today","Bhavin, are you up for a volleyball game tomorrow?"])
```

```python
df['encoding']=df["Message"].apply(get_sentence_embeding)
```
